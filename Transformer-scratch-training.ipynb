EVALUATION ON TEST DATA

```python
def evaluate_model_on_test(
    model_core, 
    test_dataloader, 
    loss_fn, 
    bleu_metric, 
    chrf_metric, 
    config, 
    tokenizer_src, 
    tokenizer_tgt, 
    device
):
    """
    Runs a full evaluation on the test dataset (batch_size=1).
    It calculates BOTH loss (using teacher forcing) and
    metrics (using greedy_decode) in a single pass.
    """
    model_core.eval()
    
    total_loss = 0.0
    count = 0
    
    # Ensure metrics are reset before evaluation
    bleu_metric.reset()
    chrf_metric.reset()
    
    # Create an iterator for the test set
    test_iterator = tqdm(test_dataloader, desc="Running Test Evaluation")
    
    with torch.no_grad():
        for batch in test_iterator:
            if batch is None: continue
            count += 1
            
            # --- (1) Calculate Loss (Teacher Forcing) ---
            
            if config['model_type'] == 'scratch':
                encoder_input = batch['encoder_input'].to(device, non_blocking=True)
                decoder_input = batch['decoder_input'].to(device, non_blocking=True)
                encoder_mask = batch['scratch_encoder_mask'].to(device, non_blocking=True)
                decoder_mask = batch['scratch_decoder_mask'].to(device, non_blocking=True)
                
                encoder_output = model_core.encode(encoder_input, encoder_mask)
                decoder_output = model_core.decode(encoder_output, encoder_mask, decoder_input, decoder_mask)
            
            else: # 'pytorch'
                encoder_input = batch['encoder_input'].to(device, non_blocking=True)
                decoder_input = batch['decoder_input'].to(device, non_blocking=True)
                src_padding_mask = batch['src_padding_mask'].to(device, non_blocking=True)
                tgt_padding_mask = batch['tgt_padding_mask'].to(device, non_blocking=True)
                tgt_seq_len = decoder_input.size(1)
                tgt_causal_mask = nn.Transformer.generate_square_subsequent_mask(tgt_seq_len).to(device, non_blocking=True)
                
                encoder_output = model_core.encode(encoder_input, src_padding_mask)
                decoder_output = model_core.decode(encoder_output, src_padding_mask, decoder_input, tgt_padding_mask, tgt_causal_mask)
            
            proj_output = model_core.project(decoder_output)
            label = batch['label'].to(device, non_blocking=True)
            
            loss = loss_fn(proj_output.view(-1, tokenizer_tgt.get_vocab_size()), label.view(-1))
            total_loss += loss.item()
            
            
            # --- (2) Calculate Metrics (Greedy Decode) ---
            
            # We must use the correct mask for greedy_decode (scratch_encoder_mask)
            encoder_mask_for_greedy = batch['scratch_encoder_mask'].to(device, non_blocking=True)
            
            model_out_ids = greedy_decode(
                model_core, 
                encoder_input, 
                encoder_mask_for_greedy, 
                config, 
                tokenizer_tgt, 
                config['seq_len_tgt'], 
                device
            )

            # Get text representations
            target_text = batch["tgt_text"][0]
            model_out_text = tokenizer_tgt.decode(model_out_ids.detach().cpu().tolist())
            
            pred_text_list = [model_out_text]
            ref_text_list = [[target_text]]
            
            # Update metrics
            bleu_metric.update(pred_text_list, ref_text_list)
            chrf_metric.update(pred_text_list, ref_text_list)
            
            # Update progress bar
            test_iterator.set_postfix({
                "current_loss": f"{loss.item():6.3f}",
                "avg_loss": f"{total_loss/count:6.3f}"
            })

    if count == 0:
        print("Warning: Test dataloader was empty.")
        return 0.0, 0.0, 0.0

    # Compute final results
    final_avg_loss = total_loss / count
    final_bleu = bleu_metric.compute()
    final_chrf = chrf_metric.compute()
    
    return final_avg_loss, final_bleu, final_chrf
```

MAIN EXECUTION SCRIPT

```python
def main_test(config):
    # --- CHOOSE YOUR EXPERIMENT ---
    # ------------------------------
    # To test scratch model
    config = get_scratch_config() 

    # To test the PyTorch model
    # config = get_pytorch_config() 
    # ------------------------------

    print(f"\n--- STARTING TEST RUN ---")
    print(f"Experiment: {config['experiment_name']}")
    print(f"Model Type: {config['model_type']}")

    # --- 1. Load Datasets and Tokenizers ---
    # We only need the test_dataloader and tokenizers
    # get_ds() will build tokenizers if they don't exist in /kaggle/working/
    print("Loading test dataset and tokenizers...")
    try:
        _, _, _, test_dataloader, tokenizer_src, tokenizer_tgt = get_ds(config)
    except Exception as e:
        print(f"Error loading dataset. Make sure tokenizers are built or available.")
        print(f"Error: {e}")
        return

    print(f"Test dataloader loaded with {len(test_dataloader)} samples.")
    
    # --- 2. Get Model Config and Utilities ---
    src_pad_id = tokenizer_src.token_to_id("[PAD]")
    tgt_pad_id = tokenizer_tgt.token_to_id("[PAD]")
    src_vocab_len = tokenizer_src.get_vocab_size()
    tgt_vocab_len = tokenizer_tgt.get_vocab_size()

    loss_fn = nn.CrossEntropyLoss(ignore_index=tgt_pad_id, label_smoothing=0.1).to(device)

    bleu_metric = BLEUScore().to(device)
    chrf_metric = CHRFScore().to(device)

    # --- 3. Define Models to Test ---
    # These are the two models we saved during training
    model_checkpoints = ["best_loss", "best_chrf"]
    results = {}

    # --- 4. Loop and Test Each Model ---
    for checkpoint_name in model_checkpoints:
        print(f"\n--- Testing model: {checkpoint_name} ---")
        
        # Get path to the saved weights
        model_filename = get_weights_file_path(config, checkpoint_name)
        
        if not model_filename.exists():
            print(f"Error: Model file not found at {model_filename}")
            print(f"Skipping test for {checkpoint_name}.")
            results[checkpoint_name] = {'loss': -1, 'bleu': -1, 'chrf': -1}
            continue

        # Build a fresh model instance
        model = get_model(
            config, 
            src_vocab_len, 
            tgt_vocab_len, 
            src_pad_id, 
            tgt_pad_id
        )
        model = model.to(device)

        # Load the saved weights
        print(f"Loading weights from {model_filename}...")
        state = torch.load(str(model_filename), map_location=device)
        
        # Load the state dict (this works even if training used DataParallel)
        model.load_state_dict(state['model_state_dict'])
        print(f"Weights loaded. Model was trained for {state.get('epoch', 'N/A')} epochs.")

        # Handle DataParallel if testing on multiple GPUs
        # (Same logic as in train_model)
        model_to_test = model
        if device.type == 'cuda' and torch.cuda.device_count() > 1:
            print(f"Using {torch.cuda.device_count()} GPUs for testing!")
            model = nn.DataParallel(model)
            model_to_test = model.module # We always pass the core model
        
        # Run the full evaluation
        avg_loss, bleu, chrf = evaluate_model_on_test(
            model_to_test,
            test_dataloader,
            loss_fn,
            bleu_metric,
            chrf_metric,
            config,
            tokenizer_src,
            tokenizer_tgt,
            device
        )
        
        # Store results
        results[checkpoint_name] = {
            'loss': avg_loss,
            'bleu': bleu.item(), 
            'chrf': chrf.item()
        }
        
        print(f"--- Results for {checkpoint_name} ---")
        print(f"  Test Loss: {avg_loss:6.4f}")
        print(f"  Test BLEU: {bleu.item():6.4f}")
        print(f"  Test ChrF: {chrf.item():6.4f}")

    # --- 5. Print Final Summary Report ---
    print("\n\n" + "="*30)
    print("   FINAL TEST REPORT   ")
    print("="*30)
    print(f"Experiment: {config['experiment_name']}\n")
    
    # Create a formatted table
    print("| Model         | Test Loss | Test BLEU | Test ChrF |")
    print("|---------------|-----------|-----------|-----------|")
    
    for model_name, metrics in results.items():
        print(f"| {model_name:<13} | {metrics['loss']:<9.4f} | {metrics['bleu']:<9.4f} | {metrics['chrf']:<9.4f} |")

    print("\nTest run complete.")
```

Run Test

```python
config = get_scratch_config()
main_test(config)
```
